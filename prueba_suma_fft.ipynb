## Paso 1. Importar librerías

```python
import cv2 as cv
import numpy as np
```

## Paso 2. Cargar las imágenes

```python
ojos = cv.imread("images/ojos/ho001.jpg", cv.IMREAD_GRAYSCALE)
nariz = cv.imread("images/nariz/hn004.jpg", cv.IMREAD_GRAYSCALE)
boca = cv.imread("images/boca/hb001.jpg", cv.IMREAD_GRAYSCALE)
```

## Paso 4. Redimensionar las imágenes

```python
from tools import redimenxiona_misma_forma
from matplotlib import pyplot as plt
datos = [ojos, nariz, boca]
partes_rostro = redimenxiona_misma_forma(datos)
fig, axs = plt.subplots(1, 3, figsize=(10, 5))

# Mostrar cada imagen en su respectiva subfigura
axs[0].imshow(partes_rostro[0])
axs[0].set_title('Imagen 1')

axs[1].imshow(partes_rostro[1])
axs[1].set_title('Imagen 2')

axs[2].imshow(partes_rostro[2])
axs[2].set_title('Imagen 3')
```

## Paso 5. Suma y multiplicacion de las imagenes en el dominio de la frecuencia

```python
resultado_suma = np.zeros_like(partes_rostro[0])
resultado_multiplicacion = np.ones_like(partes_rostro[0])
# Aplicar la transformada de Fourier a cada parte del rostro
for parte in partes_rostro:
    parte_fft = np.fft.fftshift(np.fft.fft2(parte))
    resultado_suma = resultado_suma + parte_fft
    resultado_multiplicacion = resultado_multiplicacion * parte_fft

resultado_suma = np.fft.ifft2(resultado_suma)
resultado_multiplicacion = np.fft.ifft2(resultado_multiplicacion)
# plt.imshow(np.abs(resultado_suma))
fig, axs = plt.subplots(1, 2, figsize=(10, 5))

# Mostrar cada imagen en su respectiva subfigura
axs[0].imshow(np.abs(resultado_suma))
axs[0].set_title('Resultado Suma')

axs[1].imshow(np.abs(resultado_multiplicacion))
axs[1].set_title('Resulltado Multiplicación')
```
